### Вариант задания 4


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Product:

• Атрибуты: 

o  Название (Name) 

o  Цена (Price) 

o  Производитель (Manufacturer)

• Методы:

o GetInfo(): метод для получения информации о продукте в виде строки.

o Discount(): метод для применения скидки к цене продукта.

o Display(): метод для отображения информации о продукте на экране.

Требования к производным классам:

1. Электроника (Electronics): Должен содержать дополнительные атрибуты,
такие как Гарантийный срок (WarrantyPeriod). Метод Discount() должен быть
переопределен для добавления логики учета гарантийного срока при
применении скидки.

2. Одежда (Clothing): Должен содержать дополнительные атрибуты, такие как
Размер (Size). Метод Display() должен быть переопределен для добавления
информации о размере при отображении информации о продукте.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public interface IReturnable
{
    void ReturnToWarehouse();
}
public interface IDeliverable
{
    void DeliverToCustomer();
}
public class Product
{
    private string name;
    private double price;
    private string manufacturer;
    private int stock;
    private double rating;
    public string Name
    {
        get { return name; }        
        set { name = value; }       
    }

    public double Price
    {
        get { return price; }
        set { price = value; }
    }

    public string Manufacturer
    {
        get { return manufacturer; }
        set { manufacturer = value; }
    }

    public int Stock
    {
        get { return stock; }
        set { stock = value; }
    }

    public double Rating
    {
        get { return rating; }
        set { rating = value; }
    }
    public bool IsAvailable
    {
        get { return stock > 0; }
    }
    public virtual string GetInfo()
    {
        return $"Описание продукта: {name}, {price}, {manufacturer}";
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public virtual void Discount()
    {
        price = price * 0.95;
        Console.WriteLine($"Скидка 5%! Новая цена: {price}");
    }
    public void AddStock(int amount)
    {
        stock += amount; 
    }

    public void RemoveStock(int amount)
    {
        if (stock >= amount)      
        {
            stock -= amount;
        }
    }

    public void Rate(double score)
    {
        rating = score;      
    }

    public string StockInfo()
    {
        return $"На складе: {stock} шт., рейтинг: {rating}";
    }
}
public class Clothing : Product
{
    public string Size { get; set; }
    public string Material { get; set; }
    public string Season { get; set; }

    public override void Display()
    {
        Console.WriteLine($"{GetInfo()} | Размер: {Size}, Материал: {Material}");
    }

    public void Wash()
    {
        Console.WriteLine($"{Name} постирана");
    }

    public bool SuitableFor(string season)
    {
        return Season == season;
    }
}

public class Electronics : Product
{
    private int period;
    public int WarrantyPeriod
    {
        get { return period; }
        set { period = value; }
    }

    public double Power { get; set; }
    public bool HasBattery { get; set; }

    public override void Discount()
    {
        if (period >= 3)         
        {
            this.Price = this.Price * 0.90;
            Console.WriteLine($"У вас увеличенная скидка в 10%! Новая цена: {this.Price}");
        }
        else                       
        {
            base.Discount();        
        }
    }

    public void PowerOn()
    {
        Console.WriteLine($"{Name} включен в сеть");
    }

    public string WarrantyInfo()
    {
        return $"Гарантия {period} года";
    }
}

public class Books : Product
{
    private string author;
    public string Author
    {
        get { return author; }
        set { author = value; }
    }

    public int Pages { get; set; }
    public string Genre { get; set; }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Автор: {author}";
    }

    public void Read()
    {
        Console.WriteLine($"Читаем {Name}...");
    }

    public string GenreInfo()
    {
        return $"Жанр: {Genre}";
    }
}

public class Smartphone : Electronics, IReturnable, IDeliverable
{
    public string Os { get; set; }
    public int StorageGb { get; set; }

    public void InstallApp(string app)
    {
        Console.WriteLine($"Установлено приложение {app} на {Name}");
    }
    public void ReturnToWarehouse()
    {
        Console.WriteLine($"{Name} возвращен на склад");
    }
    public void DeliverToCustomer()
    {
        Console.WriteLine($"{Name} доставлен клиенту");
    }
}

Product[] products = new Product[]
{
    new Clothing { Name = "Футболка", Price = 1000, Manufacturer = "Adidas", Size = "M", Material = "Хлопок", Season = "Лето", Stock = 10 },
    new Electronics { Name = "Ноутбук", Price = 50000, Manufacturer = "HP", WarrantyPeriod = 4, Power = 65, HasBattery = true, Stock = 5 },
    new Electronics { Name = "Наушники", Price = 5000, Manufacturer = "Sony", WarrantyPeriod = 1, Power = 0, HasBattery = true, Stock = 20 },
    new Books { Name = "Война и мир", Price = 800, Manufacturer = "Триумф", Author = "Л. Толстой", Pages = 1300, Genre = "Роман", Stock = 3 }
};

foreach (Product p in products)
{
    p.Display();
    p.Discount();
    Console.WriteLine(p.StockInfo());
    Console.WriteLine();
}

Smartphone phone = new Smartphone
{
    Name = "iPhone", Price = 80000, Manufacturer = "Apple", WarrantyPeriod = 2,
    Power = 20, HasBattery = true, Stock = 15, Os = "iOS", StorageGb = 256
};
phone.Display();
phone.Discount();
phone.InstallApp("Telegram");
phone.ReturnToWarehouse();
phone.DeliverToCustomer();
Console.WriteLine(phone.WarrantyInfo());
```
