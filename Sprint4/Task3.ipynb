### Вариант задания 4


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Product:

• Атрибуты: 

o  Название (Name) 

o  Цена (Price) 

o  Производитель (Manufacturer)

• Методы:

o GetInfo(): метод для получения информации о продукте в виде строки.

o Discount(): метод для применения скидки к цене продукта.

o Display(): метод для отображения информации о продукте на экране.

Требования к производным классам:

1. Электроника (Electronics): Должен содержать дополнительные атрибуты,
такие как Гарантийный срок (WarrantyPeriod). Метод Discount() должен быть
переопределен для добавления логики учета гарантийного срока при
применении скидки.

2. Одежда (Clothing): Должен содержать дополнительные атрибуты, такие как
Размер (Size). Метод Display() должен быть переопределен для добавления
информации о размере при отображении информации о продукте.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями 


<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface IProduct
{
    string GetInfo();
    void Display();
    void Discount();
}

public interface IReportable
{
    string GetReport();
}

public class Product : IProduct
{
    private string name = "";
    private double price;
    private string manufacturer = "";
    private int stockQuantity;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public double Price
    {
        get { return price; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException(nameof(value), "Цена не может быть отрицательной.");
            price = value;
        }
    }

    public string Manufacturer
    {
        get { return manufacturer; }
        set { manufacturer = value; }
    }

    public int StockQuantity
    {
        get { return stockQuantity; }
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException(nameof(value), "Количество не может быть отрицательным.");
            stockQuantity = value;
        }
    }

    public bool IsAvailable
    {
        get { return stockQuantity > 0; }
    }

    public virtual string GetInfo()
    {
        return $"Описание продукта: {name}, {price} руб., {manufacturer}";
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public virtual void Discount()
    {
        price = price * 0.95;
        Console.WriteLine($"Скидка 5%! Новая цена: {price}");
    }

    public void Restock(int amount)
    {
        if (amount <= 0)
        {
            Console.WriteLine("Количество для пополнения должно быть положительным.");
            return;
        }

        stockQuantity += amount;
        Console.WriteLine($"{name}: на складе {stockQuantity} шт.");
    }

    public void Sell(int amount)
    {
        if (amount > stockQuantity)
        {
            Console.WriteLine($"{name}: недостаточно товара на складе.");
            return;
        }

        stockQuantity -= amount;
        Console.WriteLine($"{name}: продано {amount} шт., осталось {stockQuantity} шт.");
    }
}

public class Clothing : Product
{
    private string size = "";
    private string color = "";
    private string material = "";

    public string Size
    {
        get { return size; }
        set { size = value; }
    }

    public string Color
    {
        get { return color; }
        set { color = value; }
    }

    public string Material
    {
        get { return material; }
        set { material = value; }
    }

    public override void Display()
    {
        Console.WriteLine($"{GetInfo()} | Размер: {Size}, цвет: {Color}, материал: {Material}");
    }

    public void ChangeSize(string newSize)
    {
        size = newSize;
        Console.WriteLine($"{Name}: размер изменен на {size}");
    }

    public bool IsWinter()
    {
        return material == "шерсть" || material == "пух";
    }
}

public class Electronics : Product
{
    private int period;
    private int power;
    private string model = "";

    public int WarrantyPeriod
    {
        get { return period; }
        set { period = value; }
    }

    public int Power
    {
        get { return power; }
        set { power = value; }
    }

    public string Model
    {
        get { return model; }
        set { model = value; }
    }

    public override void Discount()
    {
        if (period >= 3)
        {
            Price = Price * 0.90;
            Console.WriteLine($"У вас увеличенная скидка в 10%! Новая цена: {Price}");
        }
        else
        {
            base.Discount();
        }
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", модель: {Model}, мощность: {Power} Вт, гарантия: {WarrantyPeriod} мес.";
    }

    public void ExtendWarranty(int months)
    {
        period += months;
        Console.WriteLine($"{Name}: гарантия продлена до {period} мес.");
    }

    public bool HasLongWarranty()
    {
        return period >= 3;
    }
}

public class Books : Product, IReportable
{
    private string author = "";
    private int pages;
    private string genre = "";

    public string Author
    {
        get { return author; }
        set { author = value; }
    }

    public int Pages
    {
        get { return pages; }
        set { pages = value; }
    }

    public string Genre
    {
        get { return genre; }
        set { genre = value; }
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Автор: {author}";
    }

    string IReportable.GetReport()
    {
        return $"Отчет по книге «{Name}»: автор {Author}, жанр {Genre}, страниц {Pages}, цена {Price} руб.";
    }

    public double GetPricePerPage()
    {
        if (pages == 0)
            return 0;
        return Price / pages;
    }

    public void ChangeGenre(string newGenre)
    {
        genre = newGenre;
        Console.WriteLine($"{Name}: жанр изменен на {genre}");
    }
}

public class ProductService
{
    private readonly IProduct _product;

    public ProductService(IProduct product)
    {
        _product = product;
    }

    public void Process()
    {
        _product.Display();   
        _product.Discount();  
    }
}

IProduct[] products = new IProduct[]
{
    new Clothing    { Name = "Футболка", Price = 1000, Manufacturer = "Adidas", Size = "M", Color = "белый", Material = "хлопок", StockQuantity = 10 },
    new Electronics { Name = "Ноутбук", Price = 50000, Manufacturer = "HP", WarrantyPeriod = 4, Power = 65, Model = "Pavilion", StockQuantity = 3 },
    new Electronics { Name = "Наушники", Price = 5000, Manufacturer = "Sony", WarrantyPeriod = 1, Power = 5, Model = "WH-1000", StockQuantity = 7 },
    new Books       { Name = "Война и мир", Price = 800, Manufacturer = "Триумф", Author = "Л. Толстой", Pages = 1300, Genre = "роман", StockQuantity = 2 }
};

foreach (IProduct p in products)
{
    ProductService service = new ProductService(p); 
    service.Process();
    Console.WriteLine();
}

IReportable report = (IReportable)products[3];
Console.WriteLine(report.GetReport());

```
