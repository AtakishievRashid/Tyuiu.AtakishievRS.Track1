### Вариант задания 4


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Product:

• Атрибуты: 

o  Название (Name) 

o  Цена (Price) 

o  Производитель (Manufacturer)

• Методы:

o GetInfo(): метод для получения информации о продукте в виде строки.

o Discount(): метод для применения скидки к цене продукта.

o Display(): метод для отображения информации о продукте на экране.

Требования к производным классам:

1. Электроника (Electronics): Должен содержать дополнительные атрибуты,
такие как Гарантийный срок (WarrantyPeriod). Метод Discount() должен быть
переопределен для добавления логики учета гарантийного срока при
применении скидки.

2. Одежда (Clothing): Должен содержать дополнительные атрибуты, такие как
Размер (Size). Метод Display() должен быть переопределен для добавления
информации о размере при отображении информации о продукте.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Product
{
    private string name;
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    private double price;
    public double Price
    {
        get { return price; }
        set
        {
            if (value < 0)
                throw new Exception("Цена не может быть отрицательной!");
            price = value;
        }
    }

    private string manufacturer;
    public string Manufacturer
    {
        get { return manufacturer; }
        set { manufacturer = value; }
    }

    private string category;
    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    private int rating;
    public int Rating
    {
        get { return rating; }
        set
        {
            if (value < 1 || value > 5)
                throw new Exception("Рейтинг должен быть от 1 до 5!");
            rating = value;
        }
    }

    private int stockQuantity;
    public int StockQuantity
    {
        get { return stockQuantity; }
        set { stockQuantity = value; }
    }

    public virtual string GetInfo()
    {
        return $"Название: {name}, Цена: {price}, Производитель: {manufacturer}, Категория: {category}, Рейтинг: {rating}";
    }

    public virtual void Discount()
    {
        price = price * 0.95;
        Console.WriteLine($"Скидка 5%! Новая цена: {price}");
    }

    public virtual void Discount(double percent)
    {
        price = price * (1 - percent / 100);
        Console.WriteLine($"Скидка {percent}%! Новая цена: {price}");
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public virtual void Display(bool detailed)
    {
        if (detailed)
            Console.WriteLine($"{GetInfo()}, На складе: {stockQuantity} шт.");
        else
            Console.WriteLine($"{name} — {price}");
    }

    public bool InStock()
    {
        return stockQuantity > 0;
    }
}

public class Electronics : Product
{
    private int warrantyPeriod;
    public int WarrantyPeriod
    {
        get { return warrantyPeriod; }
        set { warrantyPeriod = value; }
    }

    public override void Discount()
    {
        if (warrantyPeriod >= 3)
        {
            this.Price = this.Price * 0.90;
            Console.WriteLine($"Увеличенная скидка 10% (гарантия {warrantyPeriod} лет)! Новая цена: {this.Price}");
        }
        else
        {
            base.Discount();
        }
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Гарантия: {warrantyPeriod} лет";
    }
}

public class Clothing : Product
{
    private string size;
    public string Size
    {
        get { return size; }
        set { size = value; }
    }

    public override void Display()
    {
        Console.WriteLine($"{GetInfo()} | Размер: {size}");
    }
}

public class Books : Product
{
    private string author;
    public string Author
    {
        get { return author; }
        set { author = value; }
    }

    private int pages;
    public int Pages
    {
        get { return pages; }
        set { pages = value; }
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Автор: {author}, Страниц: {pages}";
    }
}

public class ProductCollection<T> where T : Product
{
    private List<T> _products = new List<T>();

    public void Add(T product)
    {
        _products.Add(product);
    }

    public void Remove(T product)
    {
        _products.Remove(product);
    }

    public void DisplayAll()
    {
        foreach (var product in _products)
        {
            product.Display();
        }
    }

    public double TotalPrice()
    {
        double total = 0;
        foreach (var product in _products)
        {
            total += product.Price;
        }
        return total;
    }
}

ProductCollection<Electronics> electronics = new ProductCollection<Electronics>();
electronics.Add(new Electronics { Name = "Ноутбук", Price = 50000, Manufacturer = "HP", Category = "Техника", Rating = 5, StockQuantity = 10, WarrantyPeriod = 4 });
electronics.Add(new Electronics { Name = "Наушники", Price = 5000, Manufacturer = "Sony", Category = "Аудио", Rating = 4, StockQuantity = 25, WarrantyPeriod = 1 });

ProductCollection<Clothing> clothes = new ProductCollection<Clothing>();
clothes.Add(new Clothing { Name = "Футболка", Price = 1000, Manufacturer = "Adidas", Category = "Спорт", Rating = 4, StockQuantity = 50, Size = "M" });

ProductCollection<Books> books = new ProductCollection<Books>();
books.Add(new Books { Name = "Война и мир", Price = 800, Manufacturer = "Эксмо", Category = "Классика", Rating = 5, StockQuantity = 5, Author = "Л. Толстой", Pages = 1300 });

ProductCollection<Product> allProducts = new ProductCollection<Product>();
allProducts.Add(new Electronics { Name = "Телефон", Price = 30000, Manufacturer = "Samsung", Category = "Техника", Rating = 5, StockQuantity = 8, WarrantyPeriod = 2 });
allProducts.Add(new Clothing { Name = "Джинсы", Price = 3000, Manufacturer = "Levi's", Category = "Одежда", Rating = 4, StockQuantity = 20, Size = "L" });
allProducts.Add(new Books { Name = "1984", Price = 700, Manufacturer = "АСТ", Category = "Фантастика", Rating = 5, StockQuantity = 12, Author = "Дж. Оруэлл", Pages = 320 });

Console.WriteLine("=== Электроника ===");
electronics.DisplayAll();
Console.WriteLine($"Сумма: {electronics.TotalPrice()}");

Console.WriteLine("\n=== Одежда ===");
clothes.DisplayAll();

Console.WriteLine("\n=== Книги ===");
books.DisplayAll();

Console.WriteLine("\n=== Все продукты ===");
allProducts.DisplayAll();
```
